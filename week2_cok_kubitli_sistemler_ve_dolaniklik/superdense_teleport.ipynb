```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

def send(bits):
    qc = QuantumCircuit(2)
    qc.h(0)
    qc.cx(0, 1)
    x, y = bits
    if y:
        qc.x(0)
    if x:
        qc.z(0)
    qc.cx(0, 1)
    qc.h(0)
    p = Statevector(qc).probabilities_dict()
    k = max(p, key=p.get)
    return k[::-1], round(float(p[k]), 3)

for b in [(0, 0), (0, 1), (1, 0), (1, 1)]:
    print(b, send(b))
```

```python
from qiskit.quantum_info import partial_trace, state_fidelity

theta, phi = 1.1, 0.7

qc = QuantumCircuit(3)
qc.ry(theta, 0)
qc.rz(phi, 0)
qc.h(1)
qc.cx(1, 2)
qc.cx(0, 1)
qc.h(0)
qc.cx(1, 2)
qc.cz(0, 2)

bob = partial_trace(Statevector(qc), [0, 1])

ref = QuantumCircuit(1)
ref.ry(theta, 0)
ref.rz(phi, 0)

print(state_fidelity(bob, Statevector(ref)))
```
