```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

def win_prob(a_angle, b_angle, a_val, b_val):
    qc = QuantumCircuit(2)
    qc.h(0)
    qc.cx(0, 1)              # |Φ+⟩
    qc.ry(-2*a_angle, 0)
    qc.ry(-2*b_angle, 1)
    probs = Statevector(qc).probabilities_dict()
    return sum(p for bits, p in probs.items()
               if (int(bits[0]) ^ int(bits[1])) == (a_val & b_val))

a0, a1 = 0, np.pi/4
b0, b1 = np.pi/8, -np.pi/8

total = 0
for a_val, a_ang in [(0, a0), (1, a1)]:
    for b_val, b_ang in [(0, b0), (1, b1)]:
        p = win_prob(a_ang, b_ang, a_val, b_val)
        print(f"a={a_val} b={b_val}  P(kazanç)={p:.4f}")
        total += p

print("Ortalama:", total/4, " | cos²(π/8) =", np.cos(np.pi/8)**2)
```
