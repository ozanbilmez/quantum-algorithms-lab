```python
import time
t = time.time()
import numpy as np
print("numpy:", round(time.time() - t, 1), "sn")

t = time.time()
from qiskit.quantum_info import Statevector, partial_trace
print("qiskit:", round(time.time() - t, 1), "sn")
```

```python
def is_product_state(amps, tol=1e-9):
    a, b, c, d = amps
    return abs(a*d - b*c) < tol

def purity_of_one_qubit(amps):
    sv = Statevector(amps)
    rho = partial_trace(sv, [0])      # q0'ı at, q1 kalsın
    return rho.purity().real

states = {
    "(i)   (|00>+|01>)/√2": [1/np.sqrt(2), 1/np.sqrt(2), 0, 0],
    "(ii)  (|00>+|11>)/√2": [1/np.sqrt(2), 0, 0, 1/np.sqrt(2)],
    "(iii) ½|00>+½|01>+½|10>-½|11>": [0.5, 0.5, 0.5, -0.5],
}

for name, amps in states.items():
    print(name, "| çarpım mı:", is_product_state(amps),
          "| saflık:", round(purity_of_one_qubit(amps), 3))
```
