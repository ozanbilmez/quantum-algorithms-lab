```python
from qiskit.quantum_info import Statevector, Operator
import numpy as np

# HZH = X doğrulaması (Hadamard-Z-Hadamard = X)
H = Operator([[1, 1], [1, -1]] / np.sqrt(2))
Z = Operator([[1, 0], [0, -1]])
X = Operator([[0, 1], [1, 0]])

result = H.compose(Z).compose(H)
print(result)
print(np.allclose(result.data, X.data))   # True olmalı
```
