```python
from qiskit.quantum_info import Statevector
import numpy as np

# global faz önemsiz: |0> ile e^(i*pi/4)|0> aynı fiziksel duruma karşılık gelir
sv1 = Statevector([1, 0])
sv2 = Statevector([np.exp(1j*np.pi/4), 0])

print("sv1:", sv1.data)
print("sv2:", sv2.data)
print("Fiziksel olarak eşdeğer mi?", sv1.equiv(sv2))
```

```python
sv3 = Statevector([1/np.sqrt(2), 1/np.sqrt(2)])        # |+⟩ durumu
sv4 = Statevector([1/np.sqrt(2), -1/np.sqrt(2)])       # |−⟩ durumu (göreli faz farkı: π)

print("sv3:", sv3.data)
print("sv4:", sv4.data)
print("sv3 == sv4?", sv3.equiv(sv4))

# Born kuralına göre olasılıklar AYNI olmalı (her ikisinde de %50-%50)
print("sv3 ölçüm olasılıkları:", sv3.probabilities())
print("sv4 ölçüm olasılıkları:", sv4.probabilities())
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc_plus = QuantumCircuit(1)
qc_plus.initialize(sv3.data, 0)
qc_plus.h(0)
print("H|+> =", Statevector(qc_plus).data)   # |0> çıkmalı

qc_minus = QuantumCircuit(1)
qc_minus.initialize(sv4.data, 0)
qc_minus.h(0)
print("H|-> =", Statevector(qc_minus).data)  # |1> çıkmalı
```
